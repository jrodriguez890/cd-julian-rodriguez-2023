```python
import random
import matplotlib.pyplot as plt

puntos = 100000
p_lado = int(puntos**0.5)
separacion = 1 / (p_lado - 1)

# Genera coordenadas de puntos uniformemente distribuidos
coordenadas_x = []
coordenadas_y = []
puntos_dentro_x = []
puntos_dentro_y = []
puntos_fuera_x = []
puntos_fuera_y = []

for i in range(p_lado):
    x = i * separacion
    for j in range(p_lado):
        y = j * separacion
        coordenadas_x.append(x)
        coordenadas_y.append(y)

                # Verifica si el punto está dentro de la 1/4 parte de la circunferencia
        if x ** 2 + y ** 2 <= 1:
            puntos_dentro_x.append(x)
            puntos_dentro_y.append(y)
        else:
            puntos_fuera_x.append(x)
            puntos_fuera_y.append(y)


# Grafica los puntos dentro de la 1/4 parte de la circunferencia en azul
plt.scatter(puntos_dentro_x, puntos_dentro_y, s=0.5, color='blue')


# Encuentra los puntos fuera de la 1/4 parte de la circunferencia
#puntos_fuera_x = [x for x in coordenadas_x if x not in puntos_dentro_x]
#puntos_fuera_y = [y for y in coordenadas_y if y not in puntos_dentro_y]

# Grafica los puntos fuera de la 1/4 parte de la circunferencia en rojo
plt.scatter(puntos_fuera_x, puntos_fuera_y, s=0.5, color='red')

# Dibuja la circunferencia
theta = [i * math.pi / 2 / p_lado for i in range(p_lado)]
circunferencia_x = [math.cos(t) for t in theta]
circunferencia_y = [math.sin(t) for t in theta]
plt.plot(circunferencia_x, circunferencia_y, color='black')

plt.xlabel('Coordenada X')
plt.ylabel('Coordenada Y')
plt.title('Puntos Distribuidos y la 1/4 parte de la Circunferencia')
plt.show()
```

```python
num_points=50000 #número de puntos a generar
points = [] #lista de las características puntos
hits = 0. #Aciertos
random.seed(4) #Fija los números aleatorios generados
for _ in range(num_points): #for loop implícito
    x, y = random.random(), random.random() #generar número aleatorios homogéneos del 0 al 1
    if x*x + y*y <= 1.0: #Si los datos cumplen con la ec. de la circunferencia, entonces los puntos están dentro
        hits += 1 #contador de puntos que están dentro
        points.append((x, y, "red")) #almacenamiento de características
    else:
        points.append((x, y, "blue")) #almacenamiento de características
```
